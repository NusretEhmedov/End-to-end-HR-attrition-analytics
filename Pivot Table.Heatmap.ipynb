```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\Lala\Downloads\archive\WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

```python
df
```

## Pivot Table 1 :

```python
df['Attrition_Flag'] = (df['Attrition'] == 'Yes').astype(int)

pivot_jobsat = pd.pivot_table(
    df,
    values='Attrition_Flag',
    index='Department',
    columns='JobSatisfaction',
    aggfunc='mean') * 100

pivot_jobsat = pivot_jobsat.round(2)
```

```python
pivot_jobsat
```

# HEATMAP

```python
plt.figure(figsize=(8, 4))
sns.heatmap(
    pivot_jobsat,
    annot=True,
    fmt='.1f',
    cmap='Reds',
    cbar_kws={'label': 'Attrition Rate (%)'})
plt.title('JobSatisfaction x Department - Attrition Rate (%)')
plt.xlabel('Job Satisfaction (1=Low, 4=High)')
plt.ylabel('Department')
plt.show()
```
