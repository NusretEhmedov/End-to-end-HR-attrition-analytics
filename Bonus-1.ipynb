```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\Lala\Downloads\archive\WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

```python
churned = df[df['Attrition'] == 'Yes'].copy()
churned['ReplacementCost'] = churned['MonthlyIncome'] * 6


total_cost = churned['ReplacementCost'].sum()
avg_cost = churned['ReplacementCost'].mean()

print("Total churned employees:", len(churned))
print(f"Avg replacement cost per employee: ${avg_cost:,.2f}")
print(f"Total annual attrition cost: ${total_cost:,.2f}")
```

```python
by_dept = churned.groupby('Department')['ReplacementCost'].agg(['count', 'sum', 'mean']).round(2)
print(by_dept)
```
