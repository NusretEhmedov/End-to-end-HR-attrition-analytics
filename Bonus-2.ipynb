```python
import pandas as pd

df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
df['Attrition_Flag'] = (df['Attrition'] == 'Yes').astype(int)
df['IncomeQuartile'] = pd.qcut(df['MonthlyIncome'], 4, labels=['Q1', 'Q2', 'Q3', 'Q4'])

baseline_rate = df['Attrition_Flag'].mean()

# Target segment: OverTime=Yes AND JobSatisfaction=1 AND lowest income quartile (Q1)
mask = (
    (df['OverTime'] == 'Yes') &
    (df['JobSatisfaction'] == 1) &
    (df['IncomeQuartile'] == 'Q1')
)
segment = df[mask]

n = len(segment)
rate = segment['Attrition_Flag'].mean()

print(f'Segment size (n)        : {n}')
print(f'Segment attrition rate  : {rate:.1%}')
print(f'Company baseline rate   : {baseline_rate:.1%}')
print(f'Risk multiplier (lift)  : {rate / baseline_rate:.1f}x')
```
