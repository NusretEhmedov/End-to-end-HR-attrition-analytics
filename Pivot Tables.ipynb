## Pivot Table 1 :

```python
df['Attrition_Flag'] = (df['Attrition'] == 'Yes').astype(int)

pivot_jobsat = pd.pivot_table(
    df,
    values='Attrition_Flag',
    index='Department',
    columns='JobSatisfaction',
    aggfunc='mean') * 100

pivot_jobsat = pivot_jobsat.round(2)
```

```python
pivot_jobsat
```

## Pivot Table 2: 

```python
pivot_wlb = pd.pivot_table(
    df,
    values='Attrition_Flag',
    index='Department',
    columns='WorkLifeBalance',
    aggfunc='mean') * 100

pivot_wlb = pivot_wlb.round(2)
```

```python
pivot_wlb
```
