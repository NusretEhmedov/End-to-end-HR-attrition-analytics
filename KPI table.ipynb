## Overall attrition rate:

```python
overall_attrition_rate = df["Attrition"].eq("Yes").mean() * 100

print(f"Overall Attrition Rate: {overall_attrition_rate:.2f}%")
```

## KPI table

```python

total_employees = len(df)
total_attritions = (df['Attrition'] == 'Yes').sum()
attrition_rate = round(total_attritions / total_employees * 100, 2)

churned = df[df['Attrition'] == 'Yes']
retained = df[df['Attrition'] == 'No']


kpi_table = pd.DataFrame({
    'KPI': [
        'Total Employees',
        'Avg MonthlyIncome',
        'Avg YearsAtCompany',
        'Avg JobSatisfaction' ],
    'Churned': [
        len(churned),
        round(churned['MonthlyIncome'].mean(), 2),
        round(churned['YearsAtCompany'].mean(), 2),
        round(churned['JobSatisfaction'].mean(), 2)],
    'Retained': [
        len(retained),
        round(retained['MonthlyIncome'].mean(), 2),
        round(retained['YearsAtCompany'].mean(), 2),
        round(retained['JobSatisfaction'].mean(), 2)]})
```
